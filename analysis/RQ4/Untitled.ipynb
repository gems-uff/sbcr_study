```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ANALYSIS_PATH = '../../analysis'
INTERMEDIARY_RESULTS = f"{ANALYSIS_PATH}/intermediary_data"
DATASET1_FILE = f"{INTERMEDIARY_RESULTS}/rq4_dataset1.csv"
DATASET2_JAVA_FILE = f"{INTERMEDIARY_RESULTS}/rq4_dataset2_Java.csv"
```

```python
dataset1 = pd.read_csv(DATASET1_FILE)
dataset2_Java = pd.read_csv(DATASET2_JAVA_FILE)
```

```python
sns.set(style="whitegrid")
sns.set_context("notebook", font_scale=1.5)
bins = 21
plt.figure(figsize=(8, 5))
ax = sns.histplot(dataset1['diff'], bins=bins, stat='percent',color='gray', label="$Dataset1$", edgecolor='black', alpha=0.5)
# ax = sns.histplot(dataset2_Java['diff'], bins=bins, stat='percent', color='blue', label="$Dataset2_{Java}$", edgecolor='black', alpha=0.5)

plt.xlabel('Diff $Sim_{SBCR-MergeGen}$', fontsize=15)
plt.ylabel('Percentage of chunks', fontsize=15)

plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

plt.grid(True, linestyle='--', alpha=0.7)
# plt.xlim(-40,40)
plt.legend()
plt.show()
```
