```python
import pandas as pd
```

```python
dataset1_training = pd.read_json('data/dataset2_JavaScript_training.json')
if 'chunk_id' not in dataset1_training.columns:
    dataset1_training['chunk_id'] = dataset1_training['merge_id'].astype(str) + '-' + dataset1_training['chunk_number'].astype(str)
dataset1_training
```

```python
dataset1_testing = pd.read_json('data/dataset2_JavaScript_testing.json')
if 'chunk_id' not in dataset1_testing.columns:
    dataset1_testing['chunk_id'] = dataset1_testing['merge_id'].astype(str) + '-' + dataset1_testing['chunk_number'].astype(str)
dataset1_testing
```

```python
import pylcs
def get_lcs(text1,text2):
    res = pylcs.lcs_sequence_idx(text1, text2)
    lcs = ''.join([text2[i] for i in res if i != -1])
    return lcs
def get_gestalt(text1,text2):
    try:
        if text1 == text2:
            return 1
        return 2*len(get_lcs(text1, text2)) / (len(text1) + len(text2))
    except Exception as e:
        print(e)
        return 0
    
def get_training_similarities(chunk_id_reference):
    test_resolution = dataset1_testing[dataset1_testing['chunk_id']==chunk_id_reference].iloc[0]['all_raw_res']
    print(test_resolution)
    v1_reference = dataset1_testing[dataset1_testing['chunk_id']==chunk_id_reference].iloc[0]['all_raw_a']
    v2_reference = dataset1_testing[dataset1_testing['chunk_id']==chunk_id_reference].iloc[0]['all_raw_b']
    data = []
    for index, row in dataset1_training.iterrows():
        similarity_res = get_gestalt(test_resolution, row['all_raw_res'])
        similarity_v1 = get_gestalt(test_resolution, row['all_raw_a'])
        similarity_v2 = get_gestalt(test_resolution, row['all_raw_b'])
        data.append([row['chunk_id'], similarity_res, similarity_v1, similarity_v2])
    return data
```

```python
data = get_training_similarities("221482-5")
```

```python
df = pd.DataFrame(data, columns=['chunk_id', 'sim_res', 'sim_v1', 'sim_v2'])
df
```
